```python
# รับเฉพาะบนเครื่อง
import tensorflow as tf
import numpy as np

x = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=np.float32)
y = np.array([[0],[0],[0],[1]], dtype=np.float32)   # ทำให้เป็น shape (4,1)

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.5),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.fit(x, y, epochs=100, verbose=1)


print(model.predict(x))
print("weights, bias =", model.layers[0].get_weights())
```

```python
# รัน AI ทำ TFLite และ .h
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf

# 1. ข้อมูลสำหรับ AND Gate
x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)
y = np.array([[0], [0], [0], [1]], dtype=np.float32)

# 2. สร้างโมเดล
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),
    tf.keras.layers.Dense(1, activation="sigmoid"),
])

model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.5),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)

# 3. เทรนโมเดล และเก็บประวัติไว้ในตัวแปร history
print("--- Starting Training ---")
history = model.fit(x, y, epochs=200, verbose=1)

# --------------------------------------------------
# 4. วาดกราฟแสดง Loss ในแต่ละ Epoch
# --------------------------------------------------
plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Loss", color="blue", linewidth=2)
plt.title("Training Loss over Epochs (AND Gate)")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()
plt.show()  # แสดงหน้าต่างกราฟขึ้นมา
# 5. แสดงผลการทำนายและ weights
print("\n--- Predictions & Weights ---")
print(model.predict(x))
print("weights, bias =", model.layers[0].get_weights())

# 6. แปลงเป็น TFLite และเซฟลงไฟล์
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

with open("and_model.tflite", "wb") as f:
    f.write(tflite_model)

print("\nSaved: and_model.tflite")

# 7. แปลงเป็นไฟล์ C Header (.h)
hex_lines = [f"0x{b:02x}" for b in tflite_model]
c_array = ", ".join(hex_lines)

h_file_content = f"""// Generated TFLite Model Array for Microcontrollers
#ifndef AND_MODEL_H
#define AND_MODEL_H

const unsigned char g_and_model[] = {{
  {c_array}
}};
const unsigned int g_and_model_len = {len(tflite_model)};

#endif // AND_MODEL_H
"""

with open("and_model.h", "w") as f:
    f.write(h_file_content)

print("Saved: and_model.h")
```
