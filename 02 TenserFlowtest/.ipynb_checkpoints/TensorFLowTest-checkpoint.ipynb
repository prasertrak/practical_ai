```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
import tensorflow as tf

# 1. สร้างข้อมูลตัวอย่าง (2000 samples, 20 features)
x = np.random.randn(2000, 20).astype(np.float32)
y = (x.sum(axis=1) > 0).astype(np.int32)

# 2. สร้างและคอมไพล์โมเดล
model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(shape=(20,)),
        tf.keras.layers.Dense(16, activation="relu"),
        tf.keras.layers.Dense(2, activation="softmax"),
    ]
)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# 3. เทรนโมเดล (เก็บประวัติไว้ในตัวแปร history)
# แนะนำเพิ่ม epochs เพื่อให้เห็นแนวโน้มของ Loss ชัดเจนขึ้น (เช่น 20-30 epochs)
history = model.fit(
    x, y, epochs=30, batch_size=32, validation_split=0.2, verbose=1
)
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
import tensorflow as tf

# 1. สร้างข้อมูลตัวอย่าง (2000 samples, 20 features)
x = np.random.randn(2000, 20).astype(np.float32)
y = (x.sum(axis=1) > 0).astype(np.int32)

# 2. สร้างและคอมไพล์โมเดล
model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(shape=(20,)),
        tf.keras.layers.Dense(16, activation="relu"),
        tf.keras.layers.Dense(2, activation="softmax"),
    ]
)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# 3. เทรนโมเดล (เก็บประวัติไว้ในตัวแปร history)
# แนะนำเพิ่ม epochs เพื่อให้เห็นแนวโน้มของ Loss ชัดเจนขึ้น (เช่น 20-30 epochs)
history = model.fit(
    x, y, epochs=30, batch_size=32, validation_split=0.2, verbose=1
)
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
import tensorflow as tf

# 1. สร้างข้อมูลตัวอย่าง (2000 samples, 20 features)
x = np.random.randn(2000, 20).astype(np.float32)
y = (x.sum(axis=1) > 0).astype(np.int32)

# 2. สร้างและคอมไพล์โมเดล
model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(shape=(20,)),
        tf.keras.layers.Dense(16, activation="relu"),
        tf.keras.layers.Dense(2, activation="softmax"),
    ]
)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# 3. เทรนโมเดล (เก็บประวัติไว้ในตัวแปร history)
# แนะนำเพิ่ม epochs เพื่อให้เห็นแนวโน้มของ Loss ชัดเจนขึ้น (เช่น 20-30 epochs)
history = model.fit(
    x, y, epochs=30, batch_size=32, validation_split=0.2, verbose=1
)
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
import tensorflow as tf

# 1. สร้างข้อมูลตัวอย่าง (2000 samples, 20 features)
x = np.random.randn(2000, 20).astype(np.float32)
y = (x.sum(axis=1) > 0).astype(np.int32)

# 2. สร้างและคอมไพล์โมเดล
model = tf.keras.Sequential(
    [
        tf.keras.layers.Input(shape=(20,)),
        tf.keras.layers.Dense(16, activation="relu"),
        tf.keras.layers.Dense(2, activation="softmax"),
    ]
)

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# 3. เทรนโมเดล (เก็บประวัติไว้ในตัวแปร history)
# แนะนำเพิ่ม epochs เพื่อให้เห็นแนวโน้มของ Loss ชัดเจนขึ้น (เช่น 20-30 epochs)
history = model.fit(
    x, y, epochs=30, batch_size=32, validation_split=0.2, verbose=1
)
```
